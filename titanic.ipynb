1. Download and preprocess the data

Preprocessing steps:
- Map Sex to 0/1 so logistic regression can use it
- Fill missing Age and Fare with the training-set median so every sample can be used

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")

features = ["Pclass", "Sex", "Age", "Fare"]

age_median = train["Age"].median()
fare_median = train["Fare"].median()

def preprocess(df):
    X = df[features].copy()
    X["Sex"] = X["Sex"].map({"male": 0, "female": 1})
    X["Age"] = X["Age"].fillna(age_median)
    X["Fare"] = X["Fare"].fillna(fare_median)
    return X

X_train = preprocess(train)
y_train = train["Survived"]
X_test = preprocess(test)
```

2. Feature choice and logistic regression

Features used:
- Pclass: cabin class is related to survival
- Sex: large survival difference by gender
- Age: age may affect who was prioritized
- Fare: ticket price is related to class and resources

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
pred = model.predict(X_test)
```

3. Train on all training data, predict on the test set, and create the Kaggle submission file

```python
submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": pred
})
submission.to_csv("submission.csv", index=False)
```
